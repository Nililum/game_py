```python
from pandas import DataFrame
```

```python
data = {
    "name" : ["lee", "kim", "seo", "Park"],
    "score" : [85, 92, 76, 98]
}

df = DataFrame(data)

print(df['score'].mean())
```

```python
print("학생 수 :", len(df))
print("평균 점수 :", df['score'].mean())
print("최고 점수 :", df['score'].max())
print("최저 점수 :", df['score'].min())
```
